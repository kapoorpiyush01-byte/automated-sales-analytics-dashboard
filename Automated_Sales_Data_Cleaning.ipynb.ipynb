```python
before_rows = df.shape[0]

df = df.dropna(how='all')

after_rows = df.shape[0]

print("Rows removed:", before_rows - after_rows)
```

```python
before_rows = df.shape[0]

df = df.drop_duplicates()

after_rows = df.shape[0]

print("Duplicate rows removed:", before_rows - after_rows)
```

```python
before_rows = df.shape[0]

df = df.dropna(subset=[
    'customer_name',
    'product_name',
    'sales',
    'profit'
])

after_rows = df.shape[0]

print("Rows removed:", before_rows - after_rows)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
df[['order_date', 'ship_date']].head(10)
```

```python
df[df['order_date'].str.contains('-', na=False)][['order_date']].head(10)
```

```python
df['order_date'] = pd.to_datetime(
    df['order_date'],
    format='mixed',
    dayfirst=True
)

df['ship_date'] = pd.to_datetime(
    df['ship_date'],
    format='mixed',
    dayfirst=True
)
```

```python
df.dtypes
```

```python
df['sales'].head(10)

```

```python
df['sales'].sample(10)
```

```python
df['sales'] = pd.to_numeric(df['sales'], errors='coerce')
```

```python
df['sales'].isnull().sum()
```

```python
# Month Number
df['Month'] = df['order_date'].dt.month

# Month Name
df['Month_Name'] = df['order_date'].dt.month_name()

# Quarter
df['Quarter'] = df['order_date'].dt.quarter

# Day Name
df['Day_Name'] = df['order_date'].dt.day_name()

# Day Number
df['Day'] = df['order_date'].dt.day

# Week Number
df['Week'] = df['order_date'].dt.isocalendar().week
```

```python
df[['order_date',
    'Month',
    'Month_Name',
    'Quarter',
    'Day_Name',
    'Week']].head()
```

```python
duplicate_count = df.duplicated().sum()
print("Duplicate Rows:", duplicate_count)
```

```python
df['Shipping_Days'] = (df['ship_date'] - df['order_date']).dt.days
```

```python
df[['order_date', 'ship_date', 'Shipping_Days']].head()
```

```python
(df['Shipping_Days'] < 0).sum()
```

```python
df.describe()
```

```python
total_sales = df['sales'].sum()
print(f"Total Sales: {total_sales:,.2f}")
```

```python
total_profit = df['profit'].sum()
print(f"Total Profit: {total_profit:,.2f}")
```

```python
total_orders = df['order_id'].nunique()
print("Total Orders:", total_orders)
```

```python
total_customers = df['customer_name'].nunique()
print("Total Customers:", total_customers)
```

```python
avg_order_value = total_sales / total_orders
print(f"Average Order Value: {avg_order_value:,.2f}")
```

```python
top_products = (
    df.groupby('product_name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_products)
```

```python
top_customers = (
    df.groupby('customer_name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_customers)
```

```python
region_sales = (
    df.groupby('region')['sales']
      .sum()
      .sort_values(ascending=False)
)

print(region_sales)
```

```python
category_sales = (
    df.groupby('category')['sales']
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

```python
monthly_sales = (
    df.groupby('Month_Name')['sales']
      .sum()
)

print(monthly_sales)
```

```python
monthly_sales = (
    df.groupby(['year', 'Month_Name'])['sales']
      .sum()
      .reset_index()
)

print(monthly_sales.head())
```

```python
market_sales = (
    df.groupby('market')['sales']
      .sum()
      .sort_values(ascending=False)
)

print(market_sales)
```

```python
profit_category = (
    df.groupby('category')['profit']
      .sum()
      .sort_values(ascending=False)
)

print(profit_category)
```

```python
shipping_analysis = (
    df.groupby('ship_mode')['Shipping_Days']
      .mean()
      .sort_values()
)

print(shipping_analysis)
```

```python
segment_sales = (
    df.groupby('segment')['sales']
      .sum()
      .sort_values(ascending=False)
)

print(segment_sales)
```
